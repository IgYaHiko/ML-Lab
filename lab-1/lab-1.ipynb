```python
import numpy as np

# Set seed for reproducibility
np.random.seed(42)
```

```python
# Create a 10x10 array with random values
original_array = np.random.randint(1, 100, size=(10, 10))
print("Original 10x10 Array:")
print(original_array)
print("\n" + "="*50 + "\n")

# Slice array A (5x7) - rows 0-4, columns 0-6
A = original_array[0:5, 0:7]
print("Array A (5x7):")
print(A)
print(f"Shape of A: {A.shape}\n")

# Slice array B (7x5) - rows 0-6, columns 0-4
B = original_array[0:7, 0:5]
print("Array B (7x5):")
print(B)
print(f"Shape of B: {B.shape}\n")

# Slice array C (5x7) - rows 0-4, columns 3-9
C = original_array[0:5, 3:10]
print("Array C (5x7):")
print(C)
print(f"Shape of C: {C.shape}\n")

# Sum of A and C (same shape: 5x7)
sum_AC = A + C
print("Sum of A and C (A + C):")
print(sum_AC)
print(f"Shape of sum: {sum_AC.shape}\n")

# Multiplication of A and B (5x7) * (7x5) = (5x5)
try:
    product_AB = np.dot(A, B)
    print("Multiplication of A and B (Matrix Multiplication):")
    print(product_AB)
    print(f"Shape of product: {product_AB.shape}")
except ValueError as e:
    print(f"Error in multiplication: {e}")
```
