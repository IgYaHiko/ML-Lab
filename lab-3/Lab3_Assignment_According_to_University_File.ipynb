```python
# Import required libraries
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

sns.set_theme(style="whitegrid")
```

## 1. Univariate - Histogram

A histogram is used to visualize the distribution of one numerical variable.


```python
np.random.seed(2)
data = np.random.randint(1, 100, 50)

plt.hist(data)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()

data
```

# Titanic - Relational Plots

The university notebook uses `sns.relplot()` for relationship plots. The following variations follow that same approach.


## 1. Relationship Plot - Basic

```python
sns.relplot(
    x="age",
    y="fare",
    s=100,
    data=titanic
)
plt.show()
```

## 2. Relationship Plot using `hue`

```python
sns.relplot(
    x="age",
    y="fare",
    s=100,
    data=titanic,
    hue="sex"
)
plt.show()
```

## 3. Relationship Plot using `hue` with Survival

```python
sns.relplot(
    x="age",
    y="fare",
    s=50,
    data=titanic,
    hue="survived"
)
plt.show()
```

## 4. Relationship Plot using `hue`, `col` and `col_wrap`

```python
sns.relplot(
    x="age",
    y="fare",
    s=100,
    data=titanic,
    hue="sex",
    col="sex",
    col_wrap=2
)
plt.show()
```

## 5. Relationship Plot using Passenger Class

```python
sns.relplot(
    x="age",
    y="fare",
    s=100,
    data=titanic,
    hue="pclass",
    col="pclass",
    col_wrap=3
)
plt.show()
```

## Titanic - Joint Plot

The university notebook describes Joint Plot as **Scatter + Histogram**.


```python
sns.jointplot(
    x="age",
    y="fare",
    data=titanic,
    s=100
)
plt.show()
```

## Titanic - PairGrid

```python
# Select useful numerical variables
titanic_numeric = titanic[
    ["age", "fare", "sibsp", "parch"]
].dropna()

a = sns.PairGrid(titanic_numeric)
a.map_diag(sns.histplot)
a.map_offdiag(sns.scatterplot)
plt.show()
```

## Titanic - Heatmap

```python
# Correlation matrix for numerical Titanic variables
corr_titanic = titanic[
    ["survived", "pclass", "age", "sibsp", "parch", "fare"]
].corr()

print(corr_titanic)

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_titanic,
    annot=True,
    linewidths=0.5
)
plt.title("Correlation Heatmap - Titanic")
plt.show()
```

# Flights - Relational Plots

These use the same `sns.relplot()` style demonstrated in the university notebook.


## 1. Relationship Plot - Basic

```python
sns.relplot(
    x="year",
    y="passengers",
    s=100,
    data=flights
)
plt.show()
```

## 2. Relationship Plot using `hue`

```python
sns.relplot(
    x="year",
    y="passengers",
    s=100,
    data=flights,
    hue="month"
)
plt.show()
```

## 3. Relationship Plot using `hue` and Month Panels

```python
sns.relplot(
    x="year",
    y="passengers",
    s=50,
    data=flights,
    hue="month",
    col="month",
    col_wrap=4
)
plt.show()
```

## 4. Relationship Plot with Month as Panels

```python
sns.relplot(
    x="year",
    y="passengers",
    s=100,
    data=flights,
    hue="month",
    col="month",
    col_wrap=3
)
plt.show()
```

## 5. Relationship Plot using Year and Month

```python
sns.relplot(
    x="year",
    y="passengers",
    s=200,
    data=flights,
    hue="month",
    col="month",
    col_wrap=4
)
plt.show()
```

## Flights - Line Relationship Plot

```python
sns.relplot(
    x="year",
    y="passengers",
    data=flights,
    kind="line"
)
plt.show()
```

## Flights - Joint Plot

```python
sns.jointplot(
    x="year",
    y="passengers",
    data=flights,
    s=100
)
plt.show()
```

## Flights - PairGrid

```python
# Convert month into a numerical value for a numerical PairGrid
flights_grid = flights.copy()
flights_grid["month_number"] = flights_grid["month"].cat.codes + 1

a = sns.PairGrid(
    flights_grid[["year", "month_number", "passengers"]]
)
a.map_diag(sns.histplot)
a.map_offdiag(sns.scatterplot)
plt.show()
```

## Flights - Heatmap

```python
# Correlation matrix
corr_flights = flights_grid[
    ["year", "month_number", "passengers"]
].corr()

print(corr_flights)

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_flights,
    annot=True,
    linewidths=0.5
)
plt.title("Correlation Heatmap - Flights")
plt.show()
```
