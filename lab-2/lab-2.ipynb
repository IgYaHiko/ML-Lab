# 5. HANDLE MISSING VALUES

```python
data.isnull().sum()
```

```python
import seaborn as sns
```

```python
data_null = pd.DataFrame(data.isnull().sum())
sns.heatmap(data_null, annot=True, cmap='Blues')
```

```python

print(data.dropna(axis=0))                          # drop null rows
```

```python
print(data.dropna(axis=1))                          # drop null cols
```

```python
print(data.dropna(axis=0, how="all"))               # drop rows all-null
```

```python
print(data.dropna(axis=0, how="any", subset=['umpire1', 'umpire2']))
```

# fill missing values

```python
mean_runs = round(data['win_by_runs'].mean())

```

```python
data['win_by_runs'] = data['win_by_runs'].fillna(mean_runs)
data = data.fillna("NP")
print(data.isnull().sum())   # all zeros
```

# 6. FILTER DATA

```python
filter = data['city'] == 'Pune'
print(data[filter])
print(len(data[filter]))
```

```python
filter = data['win_by_runs'] > 100
print(data[filter])
print(len(data[filter]))
```

```python
filter1 = data['city'] == 'Pune'
filter2 = data['win_by_runs'] > 70
print(data[filter1 & filter2])
```

# 7. VALUE COUNTS & UNIQUE VALUES

```python
print(data['city'].value_counts())
print(data['winner'].value_counts())
print(len(data[data['city'] == "Delhi"]))
print(data['city'].value_counts()['Delhi'])
print(data['city'].unique())
```
