```python
##step2  Split the data in training testing ( why do we do this?)
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x,y, test_size=0.2, random_state =1)
```

```python
xtest # dueto random_state we will get same for all
```

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
xtrain=ss.fit_transform(xtrain)
xtest=ss.transform(xtest)
```

```python
xtrain
```

```python
##Step3  Build the model
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
##Step4  Fit the model ( Train the model)
```

```python
model.fit(xtrain, ytrain)
```

```python
###Step 5 Testing the model ( prediction from test data)
y_pred= model.predict(xtest)

```

```python
y_pred
```

```python
ytest
```

```python
## step 6  Evaluation
```

```python
from sklearn.metrics import r2_score
R2score = r2_score(ytest, y_pred)
R2score
```

```python
### another metric of evaluation
```

```python
from sklearn.metrics import mean_squared_error
ms_error = mean_squared_error(ytest, y_pred)
ms_error
```

```python
## for which random state we have the best fit
```

```python
scores = []
for i in range(500):
  xtrain1,xtest1,ytrain1,ytest1 = train_test_split(x,y,train_size=0.8,random_state=i)
  xtrain1=ss.fit_transform(xtrain1)
  xtest1=ss.transform(xtest1)
  model1 = LinearRegression()
  model1.fit(xtrain1,ytrain1)
  ypred1 = model1.predict(xtest1)
  r2_1 = r2_score(ytest1,ypred1)
  scores.append(r2_1)
```

```python
np.argmax(scores)
```

```python
np.max(scores)
```
