# Pandas Q.1

```python
import pandas as pd
```

# Read the csv

```python
df = pd.read_csv("iris.csv")
```

# Display first 5

```python
df.head(5)
```

# display the last 5

```python
df.tail(5)
```

## Dataset shape, col name, info, description

```python
df.shape
df.info
df.describe()   
```

## show column

```python
print(df[["sepal length", "sepal width", "petal length", "petal width"]])
```

# Mean and standard deviation 

```python
# mean
print(df[["sepal length", "sepal width", "petal length", "petal width"]].mean())
```

```python
# standard deviation 
print(df[["sepal length", "sepal width", "petal length", "petal width"]].std())
```

# Covarience and correlation

```python
# Covarience
print(df["sepal length"].cov(df["sepal width"]))
print(df["sepal length"].cov(df["petal length"]))
print(df["sepal length"].cov(df["petal width"]))
```

# Correlation

```python
print(df["sepal length"].corr(df["sepal width"]))
print(df["sepal length"].corr(df["petal length"]))
print(df["sepal length"].corr(df["petal width"]))
```

# Numpy Q.2

## Create 3X3 array containing 1 to 9 numbers

```python
import numpy as np
```

```python
arr = np.arange(1,10).reshape(3,3)
```

```python
print(arr)
```

```python
# shape
print(arr.shape)
# dimention 
print(arr.ndim)
# size 
print(arr.size)
# data type
print(arr.dtype)
```

## Mean, Median, Standard deviation, maximum, minimum

```python
print("Mean: ", np.mean(arr))
print("Median: ", np.median(arr))
print("Standard Deviation: ", np.std(arr))
print("Max: ", np.max(arr))
print("Min: ", np.min(arr))
```

## Transpose

```python
print(arr.T)
```

## Sum of each row and column

```python
print("Each row: ", np.sum(arr, axis=0))
```

```python
print("Each colum: ", np.sum(arr, axis=1))
```

# Extract second row

```python
print(arr[1])
```

# Extract colum

```python
print(arr[:,1])
```
